# Problem Set 8 and Quiz 8

An object of this material is to gauge your understanding on centroids, centers of gravity, and the Pappus-Guldinus theorems.

## Scenario 1

A conical cup with negligible thickness is filled with liquid [as shown](./scenario-01.png).

### Question S1-1

*(1 pt)*
If the cup has no top cover, how far above the apex is its centroid located?

### Question S1-2

*(3 pt)*
If the cup has a top cover also of negligible thickness,
how far above the apex is its centroid located?

### Question S1-3

*(5 pt)*
Determine $h$ so that the liquid contacts 60 \% of the surface area on the inside of the cup.

### Question S1-4

*(5 pt)*
Determine $h$ so that the liquid occupies 75 \% of the capacity of the cup.

### Question S1-5

*(10 pt)*
The cup is filled so that the liquid contacts 75 \% of the surface area on the inside of the cup.
The liquid has a specific gravity of 1.7,
and
the cup material has a density of 2000 kilograms per square meter.
If the cup has no top cover,
how far above the apex is the center of gravity of the liquid-and-cup system?
Take the density of water is 1000 kilograms per cubic meter.

### Question S1-6

*(15 pt)*
The liquid occupies 60 \% of the capacity of the cup.
The liquid has a specific gravity of 1.7,
and
the cup material has a density of 2000 kilograms per square meter.
If the cup has a top cover of negligible thickness
and made of the same material as the cup body,
how far above the apex is the center of gravity of the liquid-and-cup system?
Take the density of water is 1000 kilograms per cubic meter.

## Scenario 2

A [prototype rocket design](./scenario-02.png) comprises
a solid conical head,
a hollow cylinder,
and
a stick having a circular cross section.

### Question S2-1

*(1 pt)*

### Question S2-2

*(1 pt)*
How far from line $a$-$a$ is the centroid of the cylinder located?

### Question S2-3

*(1 pt)*
If $x = 300$ mm,
how far from the tip of the head is the centroid of the stick located?

### Question S2-4

*(5 pt)*
If $x = 210$ mm,
how far to left of the tip of the head
is the centroid of the rocket located?

### Question S2-5

*(5 pt)*
If $x = 210$ mm,
how far above the bottom fiber of the stick
is the centroid of the rocket located?

### Question S2-6

*(15 pt)*
The head, the cylinder, and the stick
have uniform densities of 570, 405, and 305 kilograms per cubic meter, respectively.
Determine $x$ so that the center of mass is along the line $a$-$a$.
