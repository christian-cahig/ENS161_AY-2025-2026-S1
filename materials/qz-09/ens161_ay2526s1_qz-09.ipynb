# Problem Set 9 and Quiz 9

This material is to gauge your understanding on
centroids, centers of gravity, the Pappus-Guldinus theorems,
area moments of inertia, the parallel-axis theorem, and radii of gyration.

## Scenario 1

*(50 points)*
Depicted in [`rocket.png`](./rocket.png) is a toy rocket design having
a solid conical head,
a hollow cylindrical body,
and
a cylindrical stick,
whose respective densities are 600, 400, and 300 kilograms per cubic meter.
Determine the stick length if the rocket's center of gravity must not lie to the left of $a$-$a$.

## Scenario 2

*(50 points)*
Consider the plate in [`plate.png`](./plate.png)
and adopt a Cartesian coordinate system where the origin is at the lower left corner of the plate.
A solid is generated by revolving the plate about the line segment bounded by $(0, 24)$ and $(24, 0)$.
Determine the volume and surface area of the generated solid.

## Scenario 3

*(50 points)*
Consider the plate in [`plate.png`](./plate.png)
and adopt a Cartesian coordinate system where the origin is at the lower left corner of the plate.
Compute:
- the moment of inertia about an axis parallel to the $x$-axis passing through the plate's centroid,
- the moment of inertia about an axis parallel to the $y$-axis passing through the plate's centroid,
- the polar moment of inertia about the plate's centroid,
  and
- the corresponding radii of gyration of the above-mentioned moments of inertia.
